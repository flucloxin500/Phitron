```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

df = pd.read_csv('enrollment_data.csv')
df

## Adding Labels to the lines
pl.xlabel('Year')
pl.ylabel('Number of Students')
pl.title('Enrollment Data Over 10 Years')

pl.plot(df['Year'],df['Programming'],label='Programming',color='#09637E',linewidth=3,linestyle='dashdot',marker='o',markersize=7)
pl.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing',color='#7AB2B2',linewidth=3,linestyle='dashed',markersize=7,marker='o')
pl.legend() 
```
