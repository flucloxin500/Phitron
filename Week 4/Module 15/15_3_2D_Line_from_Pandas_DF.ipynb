```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

df = pd.read_csv('enrollment_data.csv')
df
```

```python
pl.xlabel('Year')
pl.ylabel('Number of Students')
pl.title('Enrollment Data Over 10 Years')

pl.plot(df['Year'],df['Programming'])
```

```python
pl.xlabel('Year')
pl.ylabel('Number of Students')
pl.title('Enrollment Data Over 10 Years')

pl.plot(df['Year'],df['Programming'])
pl.plot(df['Year'],df['Digital Marketing'])
```

```python
## Adding Labels to the lines
pl.xlabel('Year')
pl.ylabel('Number of Students')
pl.title('Enrollment Data Over 10 Years')

pl.plot(df['Year'],df['Programming'],label='Programming')
pl.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')
pl.legend(loc='lower right') 
```
