```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

students = pd.read_csv('student_IQdata.csv')
students

# Relationship between Study Hour and IQ Score - Positive Correlation

pl.scatter(students['Study_Hour'], students['IQ_Score']) 
pl.xlabel('Study Hour')
pl.ylabel('IQ Score')
pl.title('Study Hour vs IQ Score')
pl.show()
```

```python
# Relationship between Shoe Size and IQ Score - No Correlation

pl.scatter(students['Shoe_Size'], students['IQ_Score']) 
pl.xlabel('Shoe Size')
pl.ylabel('IQ Score')
pl.title('Shoe Size vs IQ Score')
pl.show()
```

```python
# Relationship between Chilling Hours and IQ Score - No Correlation


pl.xlabel('Chilling Hours')
pl.ylabel('IQ Score')
pl.title('Chilling Hours vs IQ Score')
pl.scatter(students['Chilling_Hours'], students['IQ_Score'],color='red') 
```
