```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

data = {
    'Student': ['A', 'B',  'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Gender': ['M', 'F', 'F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Study_Hour': [2, 3, 4, 1, 6, 2, 4, 3, 5],
    'Attendance': [80, 90, 85, 70, 100, 75, 88, 92, 96],
    'Grade': [75, 85, 80, 90, 70, 78, 82, 88, 91],
}

df = pd.DataFrame(data)
df
```

```python
gender_group = df.groupby('Gender').size()
gender_group
```

```python
pl.bar(gender_group.index, gender_group.values,color=['red', 'orange'],edgecolor='black')
pl.xlabel('Gender')
pl.ylabel('Count')
pl.title('Gender Distribution') 
```

```python
df = pd.read_csv('student_data (1).csv')
df

status = df.groupby('CompletionStatus').size()
pl.bar(status.index, status.values, edgecolor='black',color = ['green','yellow','red'])

pl.xlabel('Completion Status')
pl.ylabel('Count')
pl.title('Course Completion Status')

pl.show()
```

```python
status = df.groupby('Location').size()
pl.bar(status.index, status.values, edgecolor='black',color = ['green','yellow','red'])

pl.xlabel('Location')
pl.ylabel('Count')
pl.title('Student Location Distribution')

pl.show()
```
