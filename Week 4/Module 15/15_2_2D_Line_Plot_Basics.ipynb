```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

hours = [2,3,4,1,5,7,3]
days = [1,2,3,4,5,6,7]

pl.plot( days , hours )
```

```python
pl.xlabel('Days')
pl.ylabel('Hours')
pl.title('Hours spent on study each day')

pl.plot( days , hours )
```

```python
hours_1 = [2,3,4,1,5,7,3]
hours_2 = [1,2,3,4,5,6,7]
days = [1,2,3,4,5,6,7]

pl.xlabel('Days')
pl.ylabel('Hours')
pl.title('Hours spent on study each day')

pl.plot( days , hours_1 )
pl.plot( days , hours_2 )
```
