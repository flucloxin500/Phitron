```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

df = pd.read_csv('Practice Day (1).csv')
df

df['Discounted_Price'] =df['Price'] - (df['Price'] * 0.1 )
df.to_csv('discounted_products.csv')

df = pd.read_csv('discounted_products.csv')
df

len(df['EmployeeID'].unique())
df.drop_duplicates(subset=['EmployeeID'], inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df
```

```python
df.dropna(subset=['Salary'], inplace=True)
df
```

```python
df[['Marks1','Marks2','Marks3']].agg(['mean','max'])
```

```python
df['Total'] = df['Marks1'] + df['Marks2'] + df['Marks3']
df
```

```python
df['Total'].std()
```

```python
pl.hist(df['Age'],color='red', edgecolor='black')
```

```python
pl.scatter(df['Age'], df['Salary'],color='red', edgecolor='black') 
pl.xlabel('Age')
pl.ylabel('Salary')
pl.title('Age vs Salary')
pl.show()
```

```python
pl.scatter(df['Price'], df['Discounted_Price'],color='red', edgecolor='black') 
pl.xlabel('Price')
pl.ylabel('Discounted Price')
pl.title('Price vs Discounted Price')
pl.show()
```
