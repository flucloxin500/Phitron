```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

student = pd.read_csv('student_dataset_complete.csv')
student
```

```python
students_marks = student[['Marks','study_hours','attendance_rate','gender']]
students_marks
```

```python
fig = px.histogram(students_marks,x='attendance_rate')
fig.show()
```

```python
fig = px.histogram(students_marks,x='attendance_rate',color='gender',nbins=5)
fig.show()
```
