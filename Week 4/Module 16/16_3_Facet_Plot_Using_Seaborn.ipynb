```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

student = pd.read_csv('student_dataset_complete.csv')
student.tail(7)
```

```python
sns.scatterplot(x='study_hours', y='test_score', data=student,hue='gender')
```

```python
sns.relplot(x='study_hours', y='test_score', data=student,hue='gender',kind='scatter')
```

```python
sns.relplot(x='study_hours', y='test_score', data=student,col='gender',kind='scatter')
```

```python
sns.relplot(x='study_hours', y='test_score', data=student , row='hostel', col='gender' , kind='scatter')
```

```python
sns.relplot(x='study_hours', y='test_score', data=student , col = 'week' , kind='scatter')
```

```python
sns.relplot(x='study_hours', y='test_score', data=student , col = 'week' , kind='scatter',col_wrap=2)
```
