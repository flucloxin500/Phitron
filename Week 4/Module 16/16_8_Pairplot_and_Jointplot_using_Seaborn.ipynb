```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

student = pd.read_csv('student_dataset_complete.csv')
student
```

```python
students_marks = student[['Marks','study_hours','attendance_rate','gender']]
students_marks
```

```python
sns.pairplot(data=students_marks , hue='gender')
```

```python
sns.pairplot(data=students_marks,kind='hist')
```

```python
sns.jointplot(data=students_marks,x='study_hours',y='Marks',hue='gender')
```

```python
sns.jointplot(data=students_marks,x='study_hours',y='Marks',kind='kde')
```
