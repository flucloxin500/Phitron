```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression,SGDRegressor
from sklearn.metrics import mean_squared_error
```

```python
#Create a Simple Dataset
hours = np.array([2, 4, 6, 8, 10]).reshape(-1,1)
marks = np.array([55, 60, 70, 80, 88])

hours, marks
```

```python
lin_reg = LinearRegression()
lin_reg.fit(hours, marks)

print("Intercept:", lin_reg.intercept_)
print("Slope:", lin_reg.coef_)
```

```python
#Compute the cost (Mean Squared Error) using a library
pred_marks = lin_reg.predict(hours)
mse_lin = mean_squared_error(marks, pred_marks)
print("Mean Squared Error (Library):", mse_lin)
```

```python
plt.scatter(hours, marks, color='blue', label='Actual Data')

plt.plot(hours, pred_marks, color='red', label='Best Fitted Line (Linear Regression)')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Linear Regression: Hours vs Marks')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Use a Gradient Descent Regressor
sgd_reg = SGDRegressor(
    max_iter=1000,learning_rate='invscaling',eta0=0.01,random_state=42)

sgd_reg.fit(hours, marks)
print("Intercept (SGD):", sgd_reg.intercept_[0])
print("Slope (SGD):", sgd_reg.coef_[0])

pred_sgd = sgd_reg.predict(hours)
mse_sgd = mean_squared_error(marks, pred_sgd)

print("Mean Squared Error (SGD):", mse_sgd)
```

```python
plt.scatter(hours, marks, color='blue', label='Actual Data')

plt.plot(hours, pred_marks, color='red', label='LinearRegression', linestyle="--")
plt.plot(hours, pred_sgd, color='green', label='SGDRegression', linestyle="-.")
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Linear Regressionvs SGDRegressor')
plt.legend()
plt.grid(True)
plt.show()
```
