```python
import numpy as np
import pandas as pd

data = [120 , 125 , 128 , 122 , 126 , 124 , 600]

df = pd.DataFrame(({'HousePrice' : data}))

mean_val = df['HousePrice'].mean()
print("Mean:", mean_val)
```

```python
median_val = df['HousePrice'].median()
print("Median:", median_val)
```

```python
df['City'] = ['Dhaka', 'Dhaka', 'Chattogram', 'Chattogram', 'Khulna', 'Khulna', 'Dhaka']
mode_val = df['City'].mode()[0]
print("Mode:", mode_val)
```
