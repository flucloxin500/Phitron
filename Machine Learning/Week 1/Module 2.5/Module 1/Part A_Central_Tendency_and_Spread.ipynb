```python
#. Marks: 50, 60, 65, 70, 75, 80, 85, 90
#Find mean, median, mode.
#Replace 90 with 900 and recompute mean. Explain the effect of outliers.

marks = [50, 60, 65, 70, 75, 80, 85, 90]
mean = sum(marks) / len(marks)
print(f"Mean: {mean}")

median = sorted(marks)[len(marks) // 2]
print(f"Median: {median}")

mode = max(set(marks), key=marks.count)
print(f"Mode: {mode}")
```

```python
#Temperatures (°C): 29, 31, 33, 33, 32, 31, 30
#Calculate variance and standard deviation.
#Which measure is easier to interpret and why?

temp = [29, 31, 33, 33, 32, 31, 30]
mean_temp = sum(temp) / len(temp)
variance = sum((x - mean_temp) ** 2 for x in temp) / len(temp)
std_dev = variance ** 0.5   
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_dev}")


#Which measure is easier to interpret and why?
```
