```python
#Cat weights (kg): 2.5, 3.0, 3.2, 3.3, 3.4, 3.5, 3.6, 3.9, 4.0, 4.5
#Find P25, P50, P75 and IQR. Mark possible outliers.

weights = [2.5, 3.0, 3.2, 3.3, 3.4, 3.5, 3.6, 3.9, 4.0, 4.5]

n = len(weights)
P25 = weights[int(0.25 * n)]
P50 = weights[int(0.50 * n)]
P75 = weights[int(0.75 * n)]
IQR = P75 - P25
print(f"P25: {P25}")
print(f"P50: {P50}")
print(f"P75: {P75}")
print(f"IQR: {IQR}")
```
