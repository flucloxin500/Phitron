```python
#1. In a survey of 100 people:
#40 like pizza (A)
#50 like burgers (B)
#20 like both (A ∩ B)
# Find P(A), P(B), P(A ∩ B), and P(A | B).

total_people = 100
likes_pizza = 40
likes_burgers = 50
likes_both = 20

# Calculate probabilities
P_A = likes_pizza / total_people
P_B = likes_burgers / total_people  
P_A_and_B = likes_both / total_people
P_A_given_B = P_A_and_B / P_B
# Print results
print(f"P(A) = {P_A:.2f}")
print(f"P(B) = {P_B:.2f}")
print(f"P(A ∩ B) = {P_A_and_B:.2f}")
print(f"P(A | B) = {P_A_given_B:.2f}")

```
