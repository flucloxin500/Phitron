```python
import pandas as pd

titanic_path = "./Titanic-Dataset.csv"
df_titanic = pd.read_csv(titanic_path)
df_titanic.head(10)
```

```python
print("Unique Values poer column: ")
df_titanic.nunique()
```

```python
print("Distributuion of Age column: ")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize = (8 , 6))
sns.histplot(df_titanic['Age'], bins = 40, kde = True)
plt.title("Distribution of Age column")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
print("Distribution of Embarked column: ")
plt.figure(figsize = (8 , 6))
sns.countplot(data=df_titanic, x = 'Embarked')
plt.title("Embarked Distribution of Titanic Passengers")
plt.xlabel("Embarked")
plt.ylabel("Count")

plt.show()
```
