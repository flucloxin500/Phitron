```python
##Import Necessary Libraries
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
sns.set(font_scale=1.1)

df = pd.read_csv("heart.csv")

df.head(10)

target_col = "HeartDisease"

numeric_cols = ["Age", "RestingBP", "Cholesterol", "FastingBS",
                "MaxHR", "Oldpeak"]

categorical_cols = ["Sex", "ChestPainType", "RestingECG",
                    "ExerciseAngina", "ST_Slope"]

print("Target:", target_col)
print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

```python
#Categorical Features vs Target
for c in categorical_cols:
    ct = pd.crosstab(df[c],df[target_col],normalize="index")
    print(f"\nProportion of heart disease within {c}")
    display(ct)
    
    ct.plot(kind="bar", stacked=True, figsize=(6, 4))
    plt.title(f"Proportion of heart disease within {c}")
    plt.xticks(rotation=0)
    plt.ylabel("Proportion")
    plt.tight_layout()
    plt.show()
```
