```python
import pandas as pd

data =pd.DataFrame( {
    'Name': ['Walid', 'Rasim', 'Fahim', 'Nakib'],
    'Department': ['Editing' , 'Grid' , 'Software' , 'Marketing'],
    'Salary': [45000, 100000, 50000, 45000]
})

data
data.sort_values(by='Salary', ascending=True, inplace=True)
print(data)
data.sort_values(by='Name' , ascending=False, inplace=True)
print(data)
```
