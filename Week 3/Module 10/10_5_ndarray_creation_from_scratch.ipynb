### ndarray from scratch


```python
import numpy as np

# 3d Array
arr3 = np.array([
    #1st floor
    [ [1,2,3], # 1st row
    [4,5,6] ], # 2nd row
    
    #2nd floor
    [ [7,8,9], # 1st row
    [10,11,12]], # 2nd row
    
    #3rd floor
     [[7,8,9], # 1st row
    [10,11,12]] # 2nd row
    
     ])


#np.zeros

arr = np.zeros((2,3),dtype=np.int8)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.zeros_like(arr3)
print(arr)
print(arr.shape)
```

```python
#np.ones

# 3d Array
arr3 = np.array([
    #1st floor
    [ [1,2,3], # 1st row
    [4,5,6] ], # 2nd row
    
    #2nd floor
    [ [7,8,9], # 1st row
    [10,11,12]], # 2nd row
    
    #3rd floor
     [[7,8,9], # 1st row
    [10,11,12]] # 2nd row
    
     ])


#np.zeros

arr = np.ones((2,3),dtype=np.int8)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.ones_like(arr3)
print(arr)
print(arr.shape)
```

```python
#np.empty

arr = np.empty((4,3),dtype=np.int8)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty_like(arr3)
print(arr)
print(arr.shape)
```

```python
#np.full

arr = np.full((4,3),np.inf)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.full_like(arr3,np.inf,dtype=np.float64)
print(arr)
print(arr.shape)
```
