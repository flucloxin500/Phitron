#### Array creation with range function


```python
#np.arange(first,end,step) -> SImilar to range function
import numpy as np
arr = np.arange(1, 10, 1).reshape(3,3)


print(arr)
print(type(arr))
print(arr.shape) 
print(arr.dtype)
print(arr.ndim)
print(arr.size)
```

```python
#np.linespace(start , end(included) , n )

arr = np.linspace(1, 10, 10)


print(arr)
print(type(arr))
print(arr.shape)
print(arr.dtype)
print(arr.ndim)
print(arr.size)
```
