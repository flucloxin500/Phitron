```python
import numpy as np

#1d array
arr1 = np.array([1,2,3,4,5])

arr1_mod = arr1[1:4].copy()
arr1[1] = 2 

arr1_mod[2] = 200

print(arr1_mod)
print(arr1)

arr1_mod[0] = 100
print(arr1_mod)
```
