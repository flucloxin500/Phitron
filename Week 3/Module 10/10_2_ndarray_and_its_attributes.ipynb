#### NdArray

```python
import numpy as np

# 1d array
arr1 = np.array([1,2,3,4,5])

print(type(arr1))
print(arr1.ndim)

#2d Array
arr2 = np.array([ [1,2,3],
                 [4,5,6]])
print(type(arr2))
print(arr2.ndim)

# 3d Array
arr3 = np.array([
    #1st floor
    [ [1,2,3], # 1st row
    [4,5,6] ], # 2nd row
    
    #2nd floor
    [ [7,8,9], # 1st row
    [10,11,12]], # 2nd row
    
    #3rd floor
     [[7,8,9], # 1st row
    [10,11,12]] # 2nd row
    
     ])


print(type(arr3 ))
print(arr3.ndim )
```

### Array Attributes


```python
#dimensions
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

#shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#datatype
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```
