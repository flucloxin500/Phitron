### Creating Matrix for linear Algebra

```python
import numpy as np

diagonal_matrix = np.diag([1,1,1,1,1])

print (diagonal_matrix)
print(diagonal_matrix.shape) 
```

##### Identity Matrix


```python
identity_mat = np.eye(3)
print(identity_mat)
print(identity_mat.shape)

#np.eye(row , column)

mat = np.eye(3,4)
print(mat)

mat = np.eye(3,4,-1)
print(mat)

mat = np.eye(3,4,2)
print(mat)
```
