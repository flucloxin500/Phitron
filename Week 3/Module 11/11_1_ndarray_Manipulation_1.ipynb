```python
import numpy as np

arr = np.random.randint(1, 100, size=(10,5))  # Sample - 10 , features = 5
arr.shape

b = arr.reshape(5, 10)  # Reshaping to 5 , 10
b.shape
b.ndim
print(b)

flatten = b.flatten()  # Flattening the array
flatten.shape
flatten.ndim
print(flatten)

col_wise = np.ravel(b , order = 'F')  # Converting to 1D using ravel

col_wise = np.ravel(b , order = 'F')  # Converting to 1D using ravel
print(col_wise)
```

```python
a = np.random.randint(1, 10, size=(2,3)) 
b = np.random.randint(20, 30, size=(2,3)) 

print(a)
print(b)
```

### Concatenate

```python
con_row = np.concatenate((a,b), axis=0)  # Concatenating along rows
print(con_row)

con_col = np.concatenate((a,b), axis=1)  # Concatenating along rows
print(con_col)
```

```python
x = np.random.randint(1, 10, size=(2,3)) 
y = np.random.randint(20, 30, size=(1,3)) 

r = np.concatenate((x,y), axis=0)  # Row wise concatenation
print(r)
c = np.concatenate((x,y), axis=1)  # Column wise concatenation
print(c)
```
