```python
import numpy as np

x = np.array([10, 25, 10, 19, 100])
print(x)

ind = np.where(x > 20 , x , 0)
print(ind)

ind = np.where(x == 19)
print(ind)
```

```python
mat = np.array([[10,3,5] , [8,4,9]])
print(mat)

ind = np.where(mat > 5)
print(ind)

arr = np.where(mat>8,mat,0)
print(arr)
```

```python
maxi_value_ind = np.argmax(x)
print(maxi_value_ind)

y = np.array([10, 100, 10, 19, 100])

mini_value_ind = np.argmin(y)
print(mini_value_ind)
```
