```python
import numpy as np

x = np.array([10, 25, 30, 49, 50])
y = np.array([1, 2, 3, 4, 5])

add = x + y
sub = x - y
mul = x * y
div = x / y
rem = x % y

print(add)  # Output: [10, 20, 30, 40, 50, 1, 2, 3, 4, 5]
print(sub)  # Output: [9, 18, 27, 36, 45]
print(mul)  # Output: [10, 40, 90, 160
print(div)  # Output: [10. 10. 10. 10. 10.]
print(rem)  # Output: [0, 0, 0, 0, 0]
```

# Mathematical Function


```python
# Trigonometrical functions

# sin cos tan takes values in radians

sin_val = np.sin(x)
print(sin_val)  

cos_val = np.cos(x)
print(cos_val)

deg_con = np.rad2deg(x)
print(deg_con)

rad_con = np.deg2rad(deg_con)
print(rad_con)
```

```python
# Log

base_10_log_val = np.log10(x)
print(base_10_log_val)

base_2_log_val = np.log2(x)
print(base_2_log_val)

sqrt_val = np.sqrt(x)
print(sqrt_val)
```

```python
# Different

s = np.sum(x)
print(s)

cumul = np.cumulative_sum(x)
print(cumul)
```
