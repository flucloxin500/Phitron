```python
import numpy as np

arr = np.array([1,2,3])

print(arr.dtype)

arr = np.array([1.5,2.3,3.2])

print(arr.dtype) 
```

```python
arr = np.array([1,2.5,3])
#upcasted to float
print(arr.dtype)

#upcasted to String
arr = np.array([1,2.5,3,'3'])
print(arr.dtype)
```

####  Selecting a DataType for an Array


```python
arr = np.array([1,2,300])
print(arr.dtype)

arr = arr.astype(np.int16)
print(arr.dtype)
print(arr)

#Error
#arr = np.array([1,2,3.2,'Hello'],dtype=np.float64)

arr = arr.astype(np.int32)
print(arr.dtype)

arr = np.array([1,True,3.2,'Hello'])
print(arr.dtype)
```
