### Ndarray creation from Existing data

```python
#From list
import numpy as np
lst = [10,20,30,40,50.5]

arr = np.array(lst, dtype=np.int64)
print(arr) 
print(type(arr))
print(arr.dtype)

mixed_lst = [10,20.5,'30',40,'fifty']
#mixed_arr = np.array(mixed_lst)
#print(mixed_arr)
#print(type(mixed_arr))
#print(mixed_arr.dtype)


matrix = [[1,2,3], #Row 1
          [4,5,6]] #Row 2

mixed_arr = np.array(matrix)
print(mixed_arr)
print(type(mixed_arr))
print(mixed_arr.dtype)
print(mixed_arr.shape) #2 rows, 3 columns
print(mixed_arr.ndim)  #2 dimensions
```

```python
#Tuple to array

tup = (100,200,300,400)

arr = np.array(tup,dtype=np.int32)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#Set to array
st = {1,2,3}

#Tuple can't be converted to array directly, it needs to be converted to list first
arr = np.array(list(st))
print(arr.dtype)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#Dictionary to array
dct = {'a':1,'b':2,'c':3}

keys = dct.keys()
values = dct.values()
items = dct.items()

#Keys to array
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#Values to array
arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#Items to array
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
