#### Array creation with random values


```python
#np.random.rand(shape)
import numpy as np

arr  = np.random.rand(2,3)

print(arr)
print(type(arr))
print(arr.shape)
print(arr.dtype)
print(arr.ndim)
print(arr.size)
```

```python
#Integar Values
#np.random.randint(start , end , shape)

arr = np.random.randint(1,100,(3,))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#Float Values
#np.random.randint(start , end , shape)

arr = np.random.uniform(50,100,(3,4))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
