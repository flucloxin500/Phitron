```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

student = pd.read_csv('student_dataset_complete.csv')
student.tail(7)
```

```python
#axes level plot
sns.histplot(data=student,x='attendance_rate',hue='gender',bins=7,element= 'step')
```

```python
#figure level plot
sns.displot(kind='hist',data=student,x='attendance_rate',hue='gender',bins=7)
```

```python
sns.displot(kind='hist',data=student,x='attendance_rate',hue='gender',bins=7,col='gender')
```

```python
sns.displot(kind='hist',data=student,x='study_hours',hue='gender',bins=7,col='gender')
```
