```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('final-employee-ds.csv')
df
```

### Ans to the ques 1


```python
training_hours = df["TrainingHours"].astype(float).to_numpy()

m_val = np.mean(training_hours)
std_val = np.std(training_hours)
print("Mean & Standard deviation are : ", m_val , std_val)
```

### Ans to the ques 2

```python
dept_median = df.groupby("Department")["Salary"].median()
print(dept_median)
df["Salary"] = df["Salary"].fillna(df["Department"].map(dept_median))


final = df[(df['ExperienceYears'] > 12) & (df['ProjectCount'] > 5)]['Salary'].sum()
print(final)
```

### Ans to the ques 3

```python
filt = df[
    (df['Department'].isin(['IT','Finance'])) &
    (df['Age'].between(30,45)) &
    (df['PerformanceScore'] > 88)
]

filt_sort = filt.sort_values(by=['Salary'], ascending=False)

print(filt)
```

### Ans to the ques 4

```python
df['SalaryPerHour'] = df['Salary'] / (df['TrainingHours'] * 4)

df

final = (df.sort_values(['SalaryPerHour'], ascending=False).head(5)[['Name','SalaryPerHour']])
print(final)
```
