```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl

students = pd.read_csv('student_IQdata.csv')

pl.hist(students['IQ_Score'],color='red', edgecolor='black')
pl.xlabel('IQ Score')
pl.ylabel('Frequency of Students')
pl.title('Distribution of IQ Scores')

```

```python
pl.hist(students['Study_Hour'],color='red', edgecolor='black',bins=7)
pl.xlabel('Study Hour')
pl.ylabel('Frequency of Students')
pl.title('Distribution of Study Hours')
```
