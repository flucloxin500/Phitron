```python
class Shape:
    def area(self):
        return 0
    
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    
    def area(self):
        return self.w * self.h
    
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    
    def area(self):
        pi = 3.141567
        return pi * self.r * self.r
    
shape_type = input().strip()
if shape_type == "Rectangle":
    w, h = map(int, input().split())
    shape = Rectangle(w, h)
    
elif shape_type == "Circle":
    r = int(input())
    shape = Circle(r)
    
print(f"{shape.area():.2f}")
    
```
