#### Abstraction


```python
from abc import ABC , abstractmethod

class Telephone(ABC):
    
    @abstractmethod
    def make_call(self):
        pass
    
  
        
class Sphone(Telephone):
    def make_call(self):
        print("Making a call using Sphone")
        

class iPhone(Telephone):
    def make_call(self):
        print("Making a call using iPhone")
```
