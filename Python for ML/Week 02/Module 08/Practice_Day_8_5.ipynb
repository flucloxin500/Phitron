```python
class Vehicle:
    
    def __init__(self,type,name,speed):
        self.type = type
        self.name = name
        self.speed = speed
    
    def classify(self):
        if self.speed > 100:
            print("High Speed Vehicle")
        else:
            print("Nomal Vehicle")
            
    def show_info(self):
        print(f"Type: {self.type}, Name: {self.name}, Speed: {self.speed} km/h")
class Car(Vehicle):
    
    def __init__(self,name,speed):
        super().__init__("Car",name,speed)
    
    def show_info(self):
        print(f"Type: {self.type} , Name: {self.name} , Speed: {self.speed} km/h , Predicted: {self.classify()}")

class Bike(Vehicle):
    
    def __init__(self,name,speed):
        super().__init__("Bike",name,speed)
    
    def show_info(self):
        print(f"Type: {self.type} , Name: {self.name} , Speed: {self.speed} km/h , Predicted: {self.classify()}")
        

def main():
    # Read the number of vehicles to process
    n = int(input())

    vehicles = []  # List to store vehicle objects

    # Loop to read vehicle details and create objects
    for _ in range(n):
        line = input().split()
        v_type = line[0]        # Vehicle type: Car or Bike
        name = line[1]          # Vehicle name
        speed = int(line[2])    # Vehicle speed

        if type == "Car":
            vehicle = Car(name, speed)     # Create Car object
        elif type == "Bike":
            vehicle = Bike(name, speed)    # Create Bike object
        else:
            vehicle = Vehicle(type, name, speed)  # Create generic Vehicle if needed

        vehicles.append(vehicle)  # Add to the list

    # Print confirmation for each added vehicle
    for v in vehicles:
        print(f"{v.type} Added: {v.name}")

    print("\n--- Vehicle Details ---")
    # Print details and classification for each vehicle
    for v in vehicles:
        print(v.show_info())

if __name__ == "__main__":
    main()       
        
        
```
