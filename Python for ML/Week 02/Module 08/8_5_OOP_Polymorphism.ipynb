#### Polymorphism

```python
class Camera:
    def __init__(self,name):
        self.name = name 
    
    def capture(self):
        print(f"{self.name} is capturing photo")
```

```python
class Touch_Phone(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution
    
    def capture(self):
        print("Picture is captured using Touch Phone")
    
class DSLR(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution
        
    def capture(self):
        print("Picture is captured using DSLR")

class Drone(Camera):
    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution = resolution
        
    def capture(self):
        print("Picture is captured using Drone")
        
phn = Touch_Phone("iPhone", "12MP")
dslr = DSLR("Nikon", "24MP")
drn = Drone("DJI", "48MP")

phn.capture()
dslr.capture()
drn.capture()
```
