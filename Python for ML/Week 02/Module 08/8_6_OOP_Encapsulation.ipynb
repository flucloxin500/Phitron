### Encapsulation


#Designb
class Mobile:
    def __init__(self,name,model,imei):
        self.__name= name
        self.__model = model
        self.__imei = imei
        
    def charge(self):
        print("Phone is charging.")
        
    def imei_getter(self):
        return self.__imei
    
    def model_getter(self):
        return self.__model
    
    def name_getter(self):
        return self.__name
    
    def name_setter(self,name):
        self.__name = name
    
    

#Outside the class     
iPhone = Mobile("Phone","11","347686534785342")

print(iPhone.name_getter())

iPhone.name_setter("Phitron")

print(iPhone.name_getter())
